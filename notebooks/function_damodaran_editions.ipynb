# `damodaran_editions()`

Every edition of Aswath Damodaran's (NYU Stern) country risk premium workbook in `macro.damodaran_crp_editions`, at every revision, oldest first. Every cell runs against prod, read-only, as `readonly_user`.

He posts an edition every January, with extras some years (April and July in 2026), each labelled with its "as of" date. When he re-posts an edition's file with corrections under the same date, each distinct file is a revision; `latest` marks the newest, by when it was first seen. The four country functions use these to pick what `on` and `revision` mean.

```python
from datetime import date

import polars as pl
from IPython.display import Markdown

from arora_data.damodaran_editions import CrpEdition
from arora_data.sync import damodaran_editions
from arora_data.dates import today_et
from arora_data.frames import to_df


def show(rows):
    """One result or a list of results as a Markdown table."""
    rows = [r.model_dump() for r in (rows if isinstance(rows, list) else [rows])]
    if not rows:
        return Markdown("0 rows")
    head = list(rows[0])
    lines = ["|".join(head), "|".join("---" for _ in head)]
    lines += ["|".join(str(r[c]).replace("|", "/") for c in head) for r in rows]
    return Markdown(f"{len(rows)} rows\n\n" + "\n".join(lines))
```

## Every edition and revision
`mature_erp` is his premium for a mature equity market, `us_erp` the US's (stored from 2026), `vol_multiplier` the equity-to-bond volatility ratio that turns a default spread into a country premium, and `cds_benchmark` the country CDS spreads are netted against.

```python
show(damodaran_editions())
```

## January 2026's four revisions
The same edition posted four times. Revision 2 lowered the mature premium from 4.24% to 4.23%; revision 4 is his "Updated Feb 16 for sovereign ratings" re-issue (Kenya to B3). The country functions use revision 4 unless asked for another.

```python
show([e for e in damodaran_editions() if e.edition_date == date(2026, 1, 1)])
```

## The premium and multiplier by edition
Each edition at its newest revision.

```python
(to_df(damodaran_editions(), CrpEdition).filter(pl.col("latest"))
 .select("edition_date", "mature_erp", "us_erp", "vol_multiplier", "cds_benchmark"))
```

## When a file was first seen
`first_seen_at` is in US/Eastern. It is the earliest evidence of the file (a Wayback capture or the main repo's first fetch), so for the files found late it isn't when he posted them: April and July 2026 were first seen on 2026-11-12.

```python
for e in damodaran_editions()[-6:]:
    print(e.edition_date, e.revision, e.file_name, e.first_seen_at)
```

## The edition in force today
The latest edition dated on or before today, at its newest revision: what the country functions read by default.

```python
today = today_et()
print(max((e for e in damodaran_editions() if e.latest and e.edition_date <= today), key=lambda e: e.edition_date))
```
